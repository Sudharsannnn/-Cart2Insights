Understand the Dataset & ER Diagram

customers
Purpose: Contains customer information.
Primary Key: customer_id

orders
Purpose: Contains order-level information.
Primary Key: order_id
Foreign Key: customer_id → customers.customer_id

order_items
Purpose: Contains products included in each order.
Composite Key: order_id + order_item_id
Foreign Keys:
    order_id → orders.order_id
    product_id → products.product_id
    seller_id → sellers.seller_id

products
Purpose: Contains product information.
Primary Key: product_id

sellers
Purpose: Contains seller information.
Primary Key: seller_id

order_payments
Purpose: Contains payment information.
Foreign Key: order_id → orders.order_id

order_reviews
Purpose: Contains customer review information.
Primary Key: review_id
Foreign Key: order_id → orders.order_id

product_category
Purpose: Contains product category translations.

geolocation
Purpose: Contains Brazilian ZIP-code-level geographic information.
